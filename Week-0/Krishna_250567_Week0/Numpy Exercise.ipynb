## Numpy Indexing and Selection

Now you will be given a few matrices, and be asked to replicate the resulting matrix outputs:

```python
arr11 = np.linspace(1,25,25,dtype=int).reshape(5,5)
print(arr11)
```

```python
arr12 = arr11.reshape(-1)
arr12 = arr12[11:25]
arr12 = np.append(arr12,26)
arr12 = arr12.reshape(3,5)
arr12 = np.delete(arr12,4,axis=1)
print(arr12)
```

```python
print(arr12[1,3])
```

```python
arr13 = arr11.reshape(-1)
arr13 = arr13[1:12:5]
arr13 = arr13.reshape(3,1)
print(arr13)
```

```python
arr14 = arr11[4]
print(arr14)
```

```python
arr15 = arr11[3:5]
print(arr15)
```
